```r
lf <- polars:::PlRLazyFrame$new_from_csv("dummydata.csv", separator=",", has_header=TRUE,
  ignore_errors=FALSE, skip_rows=0L, cache=TRUE,
  missing_utf8_is_empty_string=FALSE, low_memory=FALSE,
  rechunk=FALSE, skip_rows_after_header=0L, encoding="utf8",
  try_parse_dates=FALSE, eol_char="\n", raise_if_empty=TRUE,
  truncate_ragged_lines=FALSE, decimal_comma=FALSE,
  glob=TRUE, row_index_offset=0L)

lf$collect_schema()
```

```r
grl_read_csv <- function(input, sep = ",", mode = c("lazy", "chunk", "full"), header = TRUE) {
  mode <- match.arg(mode)
  
  switch(mode,
    lazy = {
      lf <- polars:::PlRLazyFrame$new_from_csv(input, separator = sep, has_header = header)
      attr(lf, "schema") <- names(lf$collect_schema())
      lf
    },
    full = {
      lf <- polars:::PlRLazyFrame$new_from_csv(input, separator = sep, has_header = header)
      df <- lf$collect()
      attr(df, "schema") <- names(lf$collect_schema())
      df
    },
    chunk = {
      lf <- polars:::PlRLazyFrame$new_from_csv(input, separator = sep, has_header = header, rechunk = FALSE)
      df <- lf$collect()
      attr(df, "schema") <- names(lf$collect_schema())
      df
    }
  )
}
```

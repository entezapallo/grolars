```r
grl_read_csv <- function(input, sep = ",", mode = c("lazy", "chunk", "full"), header = TRUE) {
  mode <- match.arg(mode)
  
  switch(mode,
    lazy = {
      lf <- polars:::PlRLazyFrame$new_from_csv(input, separator = sep, has_header = header)
      attr(lf, "schema") <- names(lf$collect_schema())
      lf
    },
    full = {
      lf <- polars:::PlRLazyFrame$new_from_csv(input, separator = sep, has_header = header)
      df <- lf$collect()
      attr(df, "schema") <- names(lf$collect_schema())
      df
    },
    chunk = {
      lf <- polars:::PlRLazyFrame$new_from_csv(input, separator = sep, has_header = header, rechunk = FALSE)
      df <- lf$collect()
      attr(df, "schema") <- names(lf$collect_schema())
      df
    }
  )
}
```
